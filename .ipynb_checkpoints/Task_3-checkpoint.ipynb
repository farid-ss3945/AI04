```python
import matplotlib.pyplot as plt
import numpy as np

study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]
ages =np.array([18, 19, 18, 20, 21, 19, 22, 20, 23, 21])
size=ages*50

xpoints=np.array(study_hours)
ypoints=np.array(scores)
plt.figure(figsize=(10,6))
plt.scatter(xpoints,ypoints,c=ages,alpha=0.8,cmap='cool',marker='8',s=size)
plt.grid()
plt.colorbar(label='Age')
plt.annotate('highest score', xy=(max(study_hours),max(scores) ), xytext=(max(study_hours)-1.7,max(scores)-10),
             arrowprops=dict(facecolor='black', shrink=0.05),
             )
plt.title('Study Hours vs Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')

fig = plt.gcf()
ax_inset = fig.add_axes([0.15, 0.6, 0.25, 0.25]) 
ax_inset.plot(study_hours[:5], scores[:5], color='purple', marker='o')
ax_inset.set_title('Zoom', fontsize=8)
ax_inset.grid()

plt.show()
```
