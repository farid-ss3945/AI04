```python
import matplotlib.pyplot as plt
import numpy as np
import random

languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]
colors=["r","b","g","y","purple"]

ypoints=np.array(languages)
xpoints=np.array(students)
bar_colors = random.sample(colors, len(ypoints))
plt.figure(figsize=(9,3))
plt.xlim(0,40)
plt.barh(ypoints,xpoints,color=bar_colors)
plt.xlabel('Students')
plt.ylabel('Languages')
plt.title('Students by programming languages')
for i in range(0,len(students)):
    if(students[i]==max(students)):
        plt.text(students[i]+0.6, i-0.1,f'{xpoints[i]}',fontweight='bold',color="red")
    else:
        plt.text(students[i]+1, i-0.1,f'{xpoints[i]}')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import random

languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]
colors=["r","b","g","y","purple"]

ypoints=np.array(students)
xpoints=np.array(languages)
bar_colors = random.sample(colors, len(ypoints))
plt.figure(figsize=(9,5))
plt.ylim(0,40)
plt.bar(xpoints,ypoints,color=bar_colors)
plt.xlabel('Languages')
plt.ylabel('Students')
plt.title('Students by programming languages')
for i in range(0,len(students)):
    if students[i] == max(students):
        plt.text(i, students[i] + 0.8, f'{students[i]}', fontweight='bold', color="red", ha='center')
    else:
        plt.text(i, students[i] + 0.8, f'{students[i]}', ha='center')
plt.show()
```
